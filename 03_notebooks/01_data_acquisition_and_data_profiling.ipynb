```python
from pathlib import Path
import zipfile
import numpy as np
import pandas as pd
```

```python
PROJECT_PATH = Path.cwd().parent
print("Project Path", PROJECT_PATH.resolve())
```

```python
RAW_PATH = PROJECT_PATH / "01_Raw_Data"
RAW_PATH.mkdir(parents=True, exist_ok=True)
print("Raw_Path", RAW_PATH.resolve())
DOCUMENTATION_PATH = PROJECT_PATH / "02_Documentation"
DOCUMENTATION_PATH.mkdir(parents=True, exist_ok=True)
print("Documentation_Path", DOCUMENTATION_PATH.resolve())
```

```python
RAW_2024_PATH = RAW_PATH / "2024"
RAW_2024_PATH.mkdir(parents=True, exist_ok=True)
print("Raw_2024_Path", RAW_2024_PATH.resolve())
RAW_2025_PATH = RAW_PATH / "2025"
RAW_2025_PATH.mkdir(parents=True, exist_ok=True)
print("Raw_2025_Path", RAW_2025_PATH.resolve())
```

```python
zip_files = list(RAW_2024_PATH.glob("*.zip"))
total_zip_file = len(zip_files)

print(f"Total Zip File: {total_zip_file}")

for file in zip_files:
    print("File Name:", file.name)
    print("File Size: ", round(file.stat().st_size/(1024 *1024),2), "MB")
    print("File Location", file.resolve())
```

```python
zip_file = zip_files[0]

with zipfile.ZipFile(zip_file, "r") as z:
    files_inside_zip = z.namelist()

    for file in files_inside_zip:
        print(file)
```

```python
EXTRACT_PATH = RAW_2024_PATH / "January"

EXTRACT_PATH.mkdir(parents=True, exist_ok=True)

print("Extract_Path", EXTRACT_PATH.resolve())
```

```python
with zipfile.ZipFile(zip_file, "r") as z:
    z.extractall(EXTRACT_PATH)
```

```python
csv_files = list(EXTRACT_PATH.rglob("*.csv"))

for files in csv_files:
    print(files)
```

```python
csv_file = csv_files[0]

df = pd.read_csv(
    csv_file,
    low_memory=False
)
```

```python
flight_dates = pd.to_datetime(
    df["FlightDate"].astype("str"),
    format= "%Y-%m-%d",
    errors= "coerce"
)

print("Min Flight Dates: ", min(flight_dates))
print("Max Flight Dates: ", max(flight_dates))
```

```python
data_profile = pd.DataFrame({
    "Metrics":[
        "Source File",
        "Rows",
        "Columns",
        "Total Duplicate",
        "Missing Cells",
        "Min Flight Date",
        "Max Flight Date"
    ],
    "Values":[
        csv_file.name,
        df.shape[0],
        df.shape[1],
        df.duplicated().sum(),
        df.isna().sum().sum(),
        min(flight_dates),
        max(flight_dates)
    ]
})

data_profile
```

```python
data_profile.to_csv(
    DOCUMENTATION_PATH / "intial_data_profile.csv",
    index = False
)

print("Initial Data Profile Saved Successfully")
```

```python
column_profile = pd.DataFrame({
    "columns": df.columns,
    "data_types": df.dtypes,
    "non_null_cell": df.notna().sum(),
    "missing cell": df.isna().sum()
})

column_profile
```

```python
column_profile.to_csv(
    DOCUMENTATION_PATH / "column_profile.csv",
    index = False
)

print("Column Profile Saved Successfully")
```

```python
missing_summary = pd.DataFrame({
    "Columns": df.columns,
    "Missing_Count": df.isna().sum(),
    "Missing_Percentage": round(df.isna().mean() * 100)
})

missing_summary
```

```python
missing_summary.to_csv(
    DOCUMENTATION_PATH / "missing_summary_report.csv",
    index = False
)

print("Missing Summary Saved Successfully")
```
