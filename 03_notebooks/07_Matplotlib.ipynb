```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
x = [1, 2, 3, 4, 5, 6, 7]
z= [2, 4, 6, 10, 6, 4, 2]
```

```python
plt.style.use('ggplot')
plt.bar(x, z, color="pink", alpha=1)
plt.xlabel("health")
plt.ylabel("money")
plt.title("LIFE", color="Green")
plt.show()
```

```python
fig, ax = plt.subplots(2, 2, figsize=(10,8))
ax[0, 0].plot(x, z, linestyle='-.', marker='o', color='red', alpha=0.5)
ax[0, 1].bar(x, z, color="pink", alpha=1)
ax[1, 0].hist([x, z], color="pink", alpha=1)
ax[1, 1].pie(x, labels=z)
plt.tight_layout()
plt.show()
```
