```python
from pathlib import Path
import zipfile
import numpy as np
import pandas as pd
```

```python
PROJECT_PATH = Path.cwd().parent
print("Project Path", PROJECT_PATH.resolve())

RAW_PATH = PROJECT_PATH / "01_Raw_Data"
RAW_PATH.mkdir(parents=True, exist_ok=True)
print("Raw_Path", RAW_PATH.resolve())
DOCUMENTATION_PATH = PROJECT_PATH / "02_Documentation"
print("Documentation_Path", DOCUMENTATION_PATH.resolve())
```

```python
RAW_2024_PATH = RAW_PATH / "2024"
print("Raw_2024_Path", RAW_2024_PATH.resolve())

EXTRACT_PATH = RAW_2024_PATH / "January"
print("Extract_Path", EXTRACT_PATH.resolve())
```

```python
csv_files = list(EXTRACT_PATH.rglob("*.csv"))
csv_file = csv_files[0]

print(csv_file.name)
```

```python
df = pd.read_csv(csv_file,
                 low_memory=False)
```

```python
df.head()
```

```python
dq = pd.DataFrame({
    "Total Rows": len(df),
    "Total Columns": len(df.columns),
    "Missing Count": df.isna().sum(),
    "Missing Percentage": df.isna().mean() * 100
})

dq
```

```python
print("Full Missing:", (dq["Missing Count"] == len(df)).sum())

print("Partial Missing:", ((dq["Missing Count"] > 0) & (dq["Missing Count"]< len(df))).sum())

print("No Missing:", (dq["Missing Count"] == 0).sum())
```

```python
data_validity = df[["Cancelled", "Diverted"]].sum()

data_validity
```

```python
quality_assessment = pd.DataFrame ({
    "Column_Name": df.columns,
    "Data_Types": df.dtypes.values,
    "Missing_Count": df.isna().sum().values
})

quality_assessment
```

```python
quality_assessment.to_csv(
    DOCUMENTATION_PATH / "data_quality_assessment.csv",
        index = False
)

print("Quality Assessment Saved Successfully")
```
